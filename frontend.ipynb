```python
# ============================================
# FRONTEND CELL 4 — JAVASCRIPT
# ============================================

import os

os.makedirs("screen_rotation_project", exist_ok=True)

javascript = r"""
function rotationMatrix(angle, direction) {

    const theta =
        direction * angle * Math.PI / 180;

    return [

        [
            Math.cos(theta),
            -Math.sin(theta)
        ],

        [
            Math.sin(theta),
            Math.cos(theta)
        ]

    ];
}


function rotate() {

    const x =
        Number(
            document.getElementById("x").value
        );

    const y =
        Number(
            document.getElementById("y").value
        );

    const angle =
        Number(
            document.getElementById("angle").value
        );

    const direction =
        Number(
            document.getElementById("direction").value
        );


    const Q =
        rotationMatrix(
            angle,
            direction
        );


    const rx =
        Q[0][0] * x +
        Q[0][1] * y;


    const ry =
        Q[1][0] * x +
        Q[1][1] * y;


    const determinant =
        Q[0][0] * Q[1][1]
        -
        Q[0][1] * Q[1][0];


    const originalDistance =
        Math.sqrt(
            x*x + y*y
        );


    const rotatedDistance =
        Math.sqrt(
            rx*rx + ry*ry
        );


    const quadraticOriginal =
        x*x + y*y;


    const quadraticRotated =
        rx*rx + ry*ry;


    document.getElementById(
        "matrix"
    ).textContent =

        "[ " +
        Q[0][0].toFixed(3) +
        "  " +
        Q[0][1].toFixed(3) +
        " ]\n" +

        "[ " +
        Q[1][0].toFixed(3) +
        "  " +
        Q[1][1].toFixed(3) +
        " ]";


    document.getElementById(
        "original"
    ).textContent =

        "(" +
        x.toFixed(2) +
        ", " +
        y.toFixed(2) +
        ")";


    document.getElementById(
        "rotated"
    ).textContent =

        "(" +
        rx.toFixed(2) +
        ", " +
        ry.toFixed(2) +
        ")";


    document.getElementById(
        "orthogonal"
    ).textContent =

        "Orthogonal: ✓ QፀQ = I";


    document.getElementById(
        "determinant"
    ).textContent =

        "det(Q): " +
        determinant.toFixed(3);


    document.getElementById(
        "distance"
    ).textContent =

        "Distance preserved: " +

        (
            Math.abs(
                originalDistance -
                rotatedDistance
            ) < 0.000001

            ? "✓ Yes"

            : "✗ No"
        );


    document.getElementById(
        "quadratic"
    ).textContent =

        "Quadratic Form: " +

        (
            Math.abs(
                quadraticOriginal -
                quadraticRotated
            ) < 0.000001

            ? "✓ Invariant"

            : "✗ Changed"
        );


    draw(
        x,
        y,
        rx,
        ry
    );
}


function draw(x, y, rx, ry) {

    const canvas =
        document.getElementById(
            "canvas"
        );

    const ctx =
        canvas.getContext("2d");


    ctx.clearRect(
        0,
        0,
        canvas.width,
        canvas.height
    );


    const cx =
        canvas.width / 2;

    const cy =
        canvas.height / 2;


    // Axes

    ctx.beginPath();

    ctx.moveTo(
        0,
        cy
    );

    ctx.lineTo(
        canvas.width,
        cy
    );

    ctx.moveTo(
        cx,
        0
    );

    ctx.lineTo(
        cx,
        canvas.height
    );

    ctx.stroke();


    const scale = 1.5;


    // Original vector

    ctx.beginPath();

    ctx.moveTo(
        cx,
        cy
    );

    ctx.lineTo(
        cx + x * scale,
        cy - y * scale
    );

    ctx.stroke();


    // Rotated vector

    ctx.beginPath();

    ctx.moveTo(
        cx,
        cy
    );

    ctx.lineTo(
        cx + rx * scale,
        cy - ry * scale
    );

    ctx.stroke();


    ctx.font =
        "14px Arial";

    ctx.fillText(
        "Original",
        cx + x * scale,
        cy - y * scale
    );


    ctx.fillText(
        "Rotated",
        cx + rx * scale,
        cy - ry * scale
    );
}


// Update angle display

document
    .getElementById("angle")
    .addEventListener(
        "input",
        function() {

            document
                .getElementById(
                    "angleValue"
                )
                .textContent =
                this.value + "°";

            rotate();
        }
    );


// Run once when page loads

rotate();
"""

with open(
    "screen_rotation_project/script.js",
    "w",
    encoding="utf-8"
) as f:

    f.write(javascript)

print("✓ script.js created")
print("✓ Frontend complete!")
```
